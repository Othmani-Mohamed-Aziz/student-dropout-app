```python

import pandas as pd
import numpy as np
from sklearn.model_selection import train_test_split

# 1. Load the dataset
df = pd.read_csv("student_dropout_raw.csv")

# 2. Validate the original target
assert "Target" in df.columns, "Target column is missing."

print("Original target distribution:")
print(df["Target"].value_counts())
print("\nOriginal dataset shape:", df.shape)

# 3. Create the binary target
# 1 = Dropout
# 0 = Graduate or Enrolled (Non-Dropout)
df["Dropout_Binary"] = (df["Target"] == "Dropout").astype(int)

print("\nBinary target distribution:")
print(df["Dropout_Binary"].value_counts())
print("\nBinary target proportions:")
print(df["Dropout_Binary"].value_counts(normalize=True).round(4))

# 4. Define Scenario A: early prediction
scenario_a_features = [
    "Marital Status",
    "Application mode",
    "Application order",
    "Course",
    "Daytime/evening attendance",
    "Previous qualification",
    "Previous qualification (grade)",
    "Nacionality",
    "Mother's qualification",
    "Father's qualification",
    "Mother's occupation",
    "Father's occupation",
    "Admission grade",
    "Displaced",
    "Educational special needs",
    "Debtor",
    "Tuition fees up to date",
    "Gender",
    "Scholarship holder",
    "Age at enrollment",
    "International",
    "Unemployment rate",
    "Inflation rate",
    "GDP"
]

# 5. Define Scenario B: prediction after the first semester
first_semester_features = [
    "Curricular units 1st sem (credited)",
    "Curricular units 1st sem (enrolled)",
    "Curricular units 1st sem (evaluations)",
    "Curricular units 1st sem (approved)",
    "Curricular units 1st sem (grade)",
    "Curricular units 1st sem (without evaluations)"
]

scenario_b_features = scenario_a_features + first_semester_features

# 6. Define categorical variables
# Integer codes are categories, not continuous measurements.
categorical_features = [
    "Marital Status",
    "Application mode",
    "Application order",
    "Course",
    "Daytime/evening attendance",
    "Previous qualification",
    "Nacionality",
    "Mother's qualification",
    "Father's qualification",
    "Mother's occupation",
    "Father's occupation",
    "Displaced",
    "Educational special needs",
    "Debtor",
    "Tuition fees up to date",
    "Gender",
    "Scholarship holder",
    "International"
]

# 7. Define numerical measurements
numeric_features = [
    "Previous qualification (grade)",
    "Admission grade",
    "Age at enrollment",
    "Unemployment rate",
    "Inflation rate",
    "GDP",
    *first_semester_features
]

# 8. Validate feature lists and data quality
assert len(scenario_a_features) == len(set(scenario_a_features))
assert len(scenario_b_features) == len(set(scenario_b_features))

missing_a = sorted(set(scenario_a_features) - set(df.columns))
missing_b = sorted(set(scenario_b_features) - set(df.columns))

assert not missing_a, f"Missing Scenario A columns: {missing_a}"
assert not missing_b, f"Missing Scenario B columns: {missing_b}"

assert set(scenario_a_features).issubset(
    set(categorical_features + numeric_features)
)
assert set(scenario_b_features).issubset(
    set(categorical_features + numeric_features)
)

assert not df[scenario_b_features].isnull().any().any(), (
    "Missing feature values detected. Handle them before training."
)
assert df["Dropout_Binary"].notnull().all()

print("\nFeature validation:")
print(f"Scenario A: {len(scenario_a_features)} features")
print(f"Scenario B: {len(scenario_b_features)} features")
print("Missing columns:", missing_a + missing_b)
print("Missing feature values:", int(df[scenario_b_features].isnull().sum().sum()))
print("Categorical features:", len(categorical_features))
print("Numeric features:", len(numeric_features))

# 9. Prepare the feature matrices and target
X_A = df[scenario_a_features].copy()
X_B = df[scenario_b_features].copy()
y = df["Dropout_Binary"].copy()

# 10. Create one shared, stratified train/test split.
# Reusing the same row indices makes Scenario A and B comparable.
train_idx, test_idx = train_test_split(
    np.arange(len(df)),
    test_size=0.20,
    random_state=42,
    stratify=y
)

X_A_train = X_A.iloc[train_idx].copy()
X_A_test = X_A.iloc[test_idx].copy()

X_B_train = X_B.iloc[train_idx].copy()
X_B_test = X_B.iloc[test_idx].copy()

y_train = y.iloc[train_idx].copy()
y_test = y.iloc[test_idx].copy()

# 11. Confirm the split
print("\nTrain/test split:")
print("Training samples:", len(train_idx))
print("Testing samples:", len(test_idx))

print("\nTraining target proportions:")
print(y_train.value_counts(normalize=True).sort_index().round(4))

print("\nTesting target proportions:")
print(y_test.value_counts(normalize=True).sort_index().round(4))

print("\nPrepared datasets:")
print("Scenario A:", X_A_train.shape, "->", X_A_test.shape)
print("Scenario B:", X_B_train.shape, "->", X_B_test.shape)

print("\nFeature preparation completed successfully.")
```

```python

# Install XGBoost if needed
%pip -q install xgboost

import time
import numpy as np
import pandas as pd

from sklearn.compose import ColumnTransformer
from sklearn.preprocessing import OneHotEncoder, StandardScaler
from sklearn.pipeline import Pipeline
from sklearn.impute import SimpleImputer

from sklearn.linear_model import LogisticRegression
from sklearn.tree import DecisionTreeClassifier
from sklearn.neighbors import KNeighborsClassifier
from sklearn.ensemble import RandomForestClassifier

from sklearn.metrics import (
    accuracy_score,
    precision_score,
    recall_score,
    f1_score,
    roc_auc_score,
    average_precision_score,
    confusion_matrix
)

from xgboost import XGBClassifier

# --------------------------------------------------
# 1. Define numeric and categorical features per scenario
# --------------------------------------------------

numeric_a = [
    col for col in numeric_features
    if col in scenario_a_features
]

categorical_a = [
    col for col in categorical_features
    if col in scenario_a_features
]

numeric_b = [
    col for col in numeric_features
    if col in scenario_b_features
]

categorical_b = [
    col for col in categorical_features
    if col in scenario_b_features
]

print("Scenario A:", len(numeric_a), "numeric +",
      len(categorical_a), "categorical features")

print("Scenario B:", len(numeric_b), "numeric +",
      len(categorical_b), "categorical features")

# --------------------------------------------------
# 2. Create preprocessing pipeline
# --------------------------------------------------

def create_preprocessor(numeric_cols, categorical_cols, scale_numeric=True):
    numeric_steps = [
        ("imputer", SimpleImputer(strategy="median"))
    ]

    if scale_numeric:
        numeric_steps.append(("scaler", StandardScaler()))

    numeric_transformer = Pipeline(numeric_steps)

    categorical_transformer = Pipeline([
        ("imputer", SimpleImputer(strategy="most_frequent")),
        ("encoder", OneHotEncoder(handle_unknown="ignore"))
    ])

    return ColumnTransformer([
        ("numeric", numeric_transformer, numeric_cols),
        ("categorical", categorical_transformer, categorical_cols)
    ])

# --------------------------------------------------
# 3. Define the five baseline models
# --------------------------------------------------

models = {
    "Logistic Regression": LogisticRegression(
        max_iter=2000,
        class_weight="balanced",
        random_state=42
    ),

    "Decision Tree": DecisionTreeClassifier(
        max_depth=8,
        class_weight="balanced",
        random_state=42
    ),

    "KNN": KNeighborsClassifier(
        n_neighbors=7,
        weights="distance"
    ),

    "Random Forest": RandomForestClassifier(
        n_estimators=300,
        class_weight="balanced",
        min_samples_leaf=2,
        random_state=42,
        n_jobs=-1
    ),

    "XGBoost": XGBClassifier(
        n_estimators=300,
        max_depth=4,
        learning_rate=0.05,
        subsample=0.8,
        colsample_bytree=0.8,
        eval_metric="logloss",
        random_state=42,
        n_jobs=-1
    )
}

# --------------------------------------------------
# 4. Train and evaluate all model/scenario combinations
# --------------------------------------------------

scenarios = {
    "Scenario A - Early": {
        "features": scenario_a_features,
        "X_train": X_A_train,
        "X_test": X_A_test,
        "numeric": numeric_a,
        "categorical": categorical_a
    },
    "Scenario B - After 1st semester": {
        "features": scenario_b_features,
        "X_train": X_B_train,
        "X_test": X_B_test,
        "numeric": numeric_b,
        "categorical": categorical_b
    }
}

trained_models = {}
results = []
prediction_details = {}

for scenario_name, scenario in scenarios.items():
    print(f"\nTraining: {scenario_name}")

    for model_name, model in models.items():

        # Scale numeric data for distance/linear models.
        # Tree models do not require numeric scaling.
        scale_numeric = model_name in [
            "Logistic Regression", "KNN"
        ]

        preprocessor = create_preprocessor(
            scenario["numeric"],
            scenario["categorical"],
            scale_numeric=scale_numeric
        )

        pipeline = Pipeline([
            ("preprocessor", preprocessor),
            ("model", model)
        ])

        start_time = time.time()

        pipeline.fit(
            scenario["X_train"],
            y_train
        )

        training_seconds = time.time() - start_time

        # Positive-class probability: Dropout = 1
        probabilities = pipeline.predict_proba(
            scenario["X_test"]
        )[:, 1]

        # Initial baseline threshold
        predictions = (probabilities >= 0.5).astype(int)

        tn, fp, fn, tp = confusion_matrix(
            y_test, predictions, labels=[0, 1]
        ).ravel()

        row = {
            "Scenario": scenario_name,
            "Model": model_name,
            "Accuracy": accuracy_score(y_test, predictions),
            "Precision": precision_score(
                y_test, predictions, zero_division=0
            ),
            "Recall": recall_score(
                y_test, predictions, zero_division=0
            ),
            "F1": f1_score(
                y_test, predictions, zero_division=0
            ),
            "ROC-AUC": roc_auc_score(y_test, probabilities),
            "PR-AUC (Average Precision)": average_precision_score(
                y_test, probabilities
            ),
            "TN": tn,
            "FP": fp,
            "FN": fn,
            "TP": tp,
            "Training Time (s)": training_seconds
        }

        results.append(row)

        key = (scenario_name, model_name)

        trained_models[key] = pipeline

        prediction_details[key] = {
            "probabilities": probabilities,
            "predictions_0.5": predictions,
            "y_test": y_test.copy()
        }

        print(
            f"  {model_name}: "
            f"ROC-AUC={row['ROC-AUC']:.4f}, "
            f"Recall={row['Recall']:.4f}, "
            f"F1={row['F1']:.4f}"
        )

# --------------------------------------------------
# 5. Display comparison table
# --------------------------------------------------

results_df = pd.DataFrame(results)

metric_columns = [
    "Accuracy",
    "Precision",
    "Recall",
    "F1",
    "ROC-AUC",
    "PR-AUC (Average Precision)"
]

results_df = results_df.sort_values(
    ["Scenario", "ROC-AUC"],
    ascending=[True, False]
).reset_index(drop=True)

print("\nBASELINE MODEL COMPARISON")
display(
    results_df.style.format({
        **{metric: "{:.4f}" for metric in metric_columns},
        "Training Time (s)": "{:.2f}"
    })
)

print("\nBest ROC-AUC model per scenario:")
display(
    results_df.loc[
        results_df.groupby("Scenario")["ROC-AUC"].idxmax(),
        ["Scenario", "Model", "ROC-AUC", "PR-AUC (Average Precision)",
         "Precision", "Recall", "F1"]
    ].reset_index(drop=True).style.format({
        "ROC-AUC": "{:.4f}",
        "PR-AUC (Average Precision)": "{:.4f}",
        "Precision": "{:.4f}",
        "Recall": "{:.4f}",
        "F1": "{:.4f}"
    })
)

# Save baseline results
results_df.to_csv("baseline_model_comparison.csv", index=False)
print("\nSaved: baseline_model_comparison.csv")
```

```python

%pip -q install xgboost

import time
import numpy as np
import pandas as pd

from sklearn.base import clone
from sklearn.model_selection import StratifiedKFold, RandomizedSearchCV
from sklearn.compose import ColumnTransformer
from sklearn.preprocessing import OneHotEncoder, StandardScaler
from sklearn.pipeline import Pipeline
from sklearn.impute import SimpleImputer

from sklearn.linear_model import LogisticRegression
from sklearn.tree import DecisionTreeClassifier
from sklearn.neighbors import KNeighborsClassifier
from sklearn.ensemble import RandomForestClassifier
from sklearn.metrics import make_scorer, average_precision_score
from xgboost import XGBClassifier

# --------------------------------------------------
# 1. Preprocessing
# --------------------------------------------------

def build_preprocessor(numeric_cols, categorical_cols, scale_numeric):
    numeric_steps = [("imputer", SimpleImputer(strategy="median"))]

    if scale_numeric:
        numeric_steps.append(("scaler", StandardScaler()))

    numeric_transformer = Pipeline(numeric_steps)

    categorical_transformer = Pipeline([
        ("imputer", SimpleImputer(strategy="most_frequent")),
        ("encoder", OneHotEncoder(handle_unknown="ignore"))
    ])

    return ColumnTransformer([
        ("numeric", numeric_transformer, numeric_cols),
        ("categorical", categorical_transformer, categorical_cols)
    ])

# --------------------------------------------------
# 2. Define models and search spaces
# --------------------------------------------------

model_spaces = {
    "Logistic Regression": (
        LogisticRegression(
            max_iter=3000,
            class_weight="balanced",
            random_state=42
        ),
        {
            "model__C": [0.01, 0.1, 1.0, 10.0, 100.0],
            "model__solver": ["lbfgs"]
        },
        True
    ),

    "Decision Tree": (
        DecisionTreeClassifier(
            class_weight="balanced",
            random_state=42
        ),
        {
            "model__max_depth": [3, 5, 8, 12, None],
            "model__min_samples_split": [2, 5, 10, 20],
            "model__min_samples_leaf": [1, 2, 5, 10],
            "model__criterion": ["gini", "entropy"]
        },
        False
    ),

    "KNN": (
        KNeighborsClassifier(),
        {
            "model__n_neighbors": [5, 7, 11, 15, 21],
            "model__weights": ["uniform", "distance"],
            "model__p": [1, 2]
        },
        True
    ),

    "Random Forest": (
        RandomForestClassifier(
            class_weight="balanced",
            random_state=42,
            n_jobs=1
        ),
        {
            "model__n_estimators": [200, 300, 500],
            "model__max_depth": [None, 8, 15, 25],
            "model__min_samples_split": [2, 5, 10],
            "model__min_samples_leaf": [1, 2, 4],
            "model__max_features": ["sqrt", 0.7]
        },
        False
    ),

    "XGBoost": (
        XGBClassifier(
            eval_metric="logloss",
            random_state=42,
            n_jobs=1
        ),
        {
            "model__n_estimators": [200, 300, 500],
            "model__max_depth": [3, 4, 6],
            "model__learning_rate": [0.03, 0.05, 0.1],
            "model__subsample": [0.8, 1.0],
            "model__colsample_bytree": [0.8, 1.0],
            "model__min_child_weight": [1, 3, 5]
        },
        False
    )
}

# --------------------------------------------------
# 3. Set up cross-validation and scoring
# --------------------------------------------------

cv = StratifiedKFold(
    n_splits=5,
    shuffle=True,
    random_state=42
)

scoring = {
    "roc_auc": "roc_auc",
    "average_precision": "average_precision",
    "precision": "precision",
    "recall": "recall",
    "f1": "f1"
}

# Use fewer trials for small spaces and more for larger ones.
n_iter_per_model = {
    "Logistic Regression": 5,
    "Decision Tree": 25,
    "KNN": 20,
    "Random Forest": 25,
    "XGBoost": 30
}

# --------------------------------------------------
# 4. Tune all models in both scenarios
# --------------------------------------------------

tuning_results = []
tuned_models = {}
tuning_started = time.time()

for scenario_name, scenario in scenarios.items():
    print(f"\n{'=' * 65}")
    print(f"TUNING {scenario_name}")
    print(f"{'=' * 65}")

    for model_name, (base_model, param_space, scale_numeric) in model_spaces.items():
        print(f"\nSearching: {model_name}")
        start = time.time()

        preprocessor = build_preprocessor(
            scenario["numeric"],
            scenario["categorical"],
            scale_numeric
        )

        pipeline = Pipeline([
            ("preprocessor", preprocessor),
            ("model", clone(base_model))
        ])

        search = RandomizedSearchCV(
            estimator=pipeline,
            param_distributions=param_space,
            n_iter=n_iter_per_model[model_name],
            scoring=scoring,
            refit="roc_auc",
            cv=cv,
            random_state=42,
            n_jobs=-1,
            verbose=0,
            return_train_score=False,
            error_score="raise"
        )

        search.fit(scenario["X_train"], y_train)

        elapsed = time.time() - start
        best_index = search.best_index_
        cv_results = search.cv_results_

        record = {
            "Scenario": scenario_name,
            "Model": model_name,
            "Best CV ROC-AUC": search.best_score_,
            "CV PR-AUC": cv_results["mean_test_average_precision"][best_index],
            "CV Precision": cv_results["mean_test_precision"][best_index],
            "CV Recall": cv_results["mean_test_recall"][best_index],
            "CV F1": cv_results["mean_test_f1"][best_index],
            "CV ROC-AUC Std": cv_results["std_test_roc_auc"][best_index],
            "Search Time (s)": elapsed,
            "Best Parameters": search.best_params_
        }

        tuning_results.append(record)
        tuned_models[(scenario_name, model_name)] = search.best_estimator_

        print(f"Best CV ROC-AUC: {search.best_score_:.4f}")
        print(f"CV PR-AUC: {record['CV PR-AUC']:.4f}")
        print(f"Best parameters: {search.best_params_}")
        print(f"Search time: {elapsed:.1f} seconds")

# --------------------------------------------------
# 5. Display cross-validation results
# --------------------------------------------------

tuning_df = pd.DataFrame(tuning_results)

tuning_df = tuning_df.sort_values(
    ["Scenario", "Best CV ROC-AUC"],
    ascending=[True, False]
).reset_index(drop=True)

print("\nCROSS-VALIDATED HYPERPARAMETER SEARCH RESULTS")

display(
    tuning_df.drop(columns=["Best Parameters"]).style.format({
        "Best CV ROC-AUC": "{:.4f}",
        "CV PR-AUC": "{:.4f}",
        "CV Precision": "{:.4f}",
        "CV Recall": "{:.4f}",
        "CV F1": "{:.4f}",
        "CV ROC-AUC Std": "{:.4f}",
        "Search Time (s)": "{:.1f}"
    })
)

print("\nBest configuration per scenario:")
display(
    tuning_df.loc[
        tuning_df.groupby("Scenario")["Best CV ROC-AUC"].idxmax(),
        ["Scenario", "Model", "Best CV ROC-AUC", "CV PR-AUC",
         "CV Precision", "CV Recall", "CV F1", "Best Parameters"]
    ].reset_index(drop=True)
)

tuning_df.to_csv("cross_validation_tuning_results.csv", index=False)

print(f"\nTotal tuning time: {(time.time() - tuning_started) / 60:.2f} minutes")
print("Saved: cross_validation_tuning_results.csv")
```

```python

import numpy as np
import pandas as pd
import matplotlib.pyplot as plt

from sklearn.base import clone
from sklearn.model_selection import StratifiedKFold, cross_val_predict
from sklearn.metrics import (
    accuracy_score,
    precision_score,
    recall_score,
    f1_score,
    roc_auc_score,
    average_precision_score,
    confusion_matrix
)

# XGBoost only
xgboost_scenarios = [
    "Scenario A - Early",
    "Scenario B - After 1st semester"
]

cv = StratifiedKFold(n_splits=5, shuffle=True, random_state=42)
threshold_grid = np.round(np.arange(0.10, 0.91, 0.01), 2)

threshold_results = []
chosen_thresholds = {}
oof_probabilities = {}

# 1. Generate out-of-fold probabilities using training data only
for scenario_name in xgboost_scenarios:
    key = (scenario_name, "XGBoost")
    data = scenarios[scenario_name]

    X_train = data["X_train"]
    y_train_local = np.asarray(y_train).ravel()

    if len(X_train) != len(y_train_local):
        raise ValueError(
            f"Training size mismatch for {scenario_name}: "
            f"{len(X_train)} rows vs {len(y_train_local)} labels."
        )

    if key not in tuned_models:
        raise KeyError(f"Model not found: {key}")

    model = clone(tuned_models[key])

    probabilities = cross_val_predict(
        model,
        X_train,
        y_train_local,
        cv=cv,
        method="predict_proba",
        n_jobs=-1
    )[:, 1]

    oof_probabilities[key] = probabilities

    for threshold in threshold_grid:
        predictions = (probabilities >= threshold).astype(int)

        tn, fp, fn, tp = confusion_matrix(
            y_train_local, predictions, labels=[0, 1]
        ).ravel()

        threshold_results.append({
            "Scenario": scenario_name,
            "Model": "XGBoost",
            "Threshold": float(threshold),
            "Accuracy": accuracy_score(y_train_local, predictions),
            "Precision": precision_score(
                y_train_local, predictions, zero_division=0
            ),
            "Recall": recall_score(
                y_train_local, predictions, zero_division=0
            ),
            "F1": f1_score(
                y_train_local, predictions, zero_division=0
            ),
            "TN": tn,
            "FP": fp,
            "FN": fn,
            "TP": tp
        })

threshold_df = pd.DataFrame(threshold_results)

# 2. Select the threshold maximizing OOF F1
# Ties: higher recall, then lower threshold
for scenario_name in xgboost_scenarios:
    subset = threshold_df[
        threshold_df["Scenario"] == scenario_name
    ].sort_values(
        ["F1", "Recall", "Threshold"],
        ascending=[False, False, True]
    )

    chosen_thresholds[(scenario_name, "XGBoost")] = float(
        subset.iloc[0]["Threshold"]
    )

selected_oof_rows = []

for (scenario_name, model_name), threshold in chosen_thresholds.items():
    row = threshold_df[
        (threshold_df["Scenario"] == scenario_name) &
        (threshold_df["Threshold"] == threshold)
    ].iloc[0]

    selected_oof_rows.append(row)

selected_oof_df = pd.DataFrame(selected_oof_rows)

print("XGBOOST — SELECTED THRESHOLDS (TRAINING OOF)")
print(
    selected_oof_df[
        ["Scenario", "Threshold", "Accuracy",
         "Precision", "Recall", "F1", "FP", "FN"]
    ].round(4).to_string(index=False)
)

# 3. Plot threshold curves for XGBoost only
for scenario_name in xgboost_scenarios:
    subset = threshold_df[
        threshold_df["Scenario"] == scenario_name
    ]

    selected = chosen_thresholds[(scenario_name, "XGBoost")]

    plt.figure(figsize=(9, 5))
    plt.plot(subset["Threshold"], subset["Precision"],
             label="Precision", linewidth=2)
    plt.plot(subset["Threshold"], subset["Recall"],
             label="Recall", linewidth=2)
    plt.plot(subset["Threshold"], subset["F1"],
             label="F1-score", linewidth=2)

    plt.axvline(
        selected,
        linestyle="--",
        label=f"Selected threshold = {selected:.2f}"
    )

    plt.title(f"{scenario_name} — XGBoost Threshold Analysis")
    plt.xlabel("Decision threshold")
    plt.ylabel("Score")
    plt.ylim(0, 1.02)
    plt.grid(alpha=0.3)
    plt.legend()
    plt.tight_layout()
    plt.show()

# 4. Evaluate selected thresholds on the held-out test set
final_threshold_results = []
y_test_local = np.asarray(y_test).ravel()

for scenario_name in xgboost_scenarios:
    key = (scenario_name, "XGBoost")
    X_test = scenarios[scenario_name]["X_test"]

    if len(X_test) != len(y_test_local):
        raise ValueError(
            f"Test size mismatch for {scenario_name}: "
            f"{len(X_test)} rows vs {len(y_test_local)} labels."
        )

    threshold = chosen_thresholds[key]
    fitted_model = tuned_models[key]

    test_probabilities = fitted_model.predict_proba(X_test)[:, 1]
    test_predictions = (test_probabilities >= threshold).astype(int)

    tn, fp, fn, tp = confusion_matrix(
        y_test_local, test_predictions, labels=[0, 1]
    ).ravel()

    final_threshold_results.append({
        "Scenario": scenario_name,
        "Model": "XGBoost",
        "Threshold": threshold,
        "Accuracy": accuracy_score(y_test_local, test_predictions),
        "Precision": precision_score(
            y_test_local, test_predictions, zero_division=0
        ),
        "Recall": recall_score(
            y_test_local, test_predictions, zero_division=0
        ),
        "F1": f1_score(
            y_test_local, test_predictions, zero_division=0
        ),
        "ROC-AUC": roc_auc_score(y_test_local, test_probabilities),
        "Average Precision": average_precision_score(
            y_test_local, test_probabilities
        ),
        "TN": tn,
        "FP": fp,
        "FN": fn,
        "TP": tp
    })

final_threshold_df = pd.DataFrame(final_threshold_results)

print("\nXGBOOST — TEST RESULTS AT SELECTED THRESHOLDS")
print(final_threshold_df.round(4).to_string(index=False))

# 5. Save results
threshold_df.to_csv("xgboost_threshold_analysis_oof.csv", index=False)
selected_oof_df.to_csv(
    "xgboost_threshold_selected_oof.csv", index=False
)
final_threshold_df.to_csv(
    "xgboost_threshold_selected_test.csv", index=False
)

print("\nSaved:")
print("- xgboost_threshold_analysis_oof.csv")
print("- xgboost_threshold_selected_oof.csv")
print("- xgboost_threshold_selected_test.csv")
```
